### <font color='#B22222'>Task 3</font> - Format plot parameters
A plot is not complete without a title, axes labels and a legend. The next example showcases how these can be added to your plot. You will also see how to change the type and/or color of the line as well as how to add a secondary y-axis to the plot. More information on how to style your plot can be found [here](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html).

If you run the code below, you should get the following output:

<img src='ancillarydata/images/exercise1/plot_task3.png'  width = 60%>

````python
# Add grid (use Seaborn's whitegrid style)
sns.set(style="whitegrid")

# Create a plot (i.e. "figure") object and set the size of your plot:
fig, ax = plt.subplots(figsize=(16, 6))

# Add plot title
# Adjust if your data frame is for another station
plt.title('Air Temperature & GPP - FLUXNET Time Series (SE-HTM)')

# Plot values for daily air temperature:
ax.plot(htm_dd_df.TIMESTAMP, htm_dd_df.TA_F,
        linestyle = '-.', linewidth = 0.6, color = 'darkorange',
        label = 'Air Temperature')

# Plot values for daytime GPP (daily values):
ax.plot(htm_dd_df.TIMESTAMP, htm_dd_df.GPP_DT_VUT_MEAN,
        linestyle = '-', linewidth = 1.7, color = '#6699CC',
        label = 'GPP')

# Add secondary y-axis:
secaxy = ax.secondary_yaxis('right')

# Add x-axis label:
plt.xlabel('Time')

# Add y-axis label:
plt.ylabel(r'Air Temperature ($^\circ$C)')

# Add secondary y-axis label:
secaxy.set_ylabel('GPP (g C m$^{-2}$ d$^{-1}$)')

# Add legend:
ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left', borderaxespad=0.)
    
# Show plot:
plt.show()
````

### <font color='#B22222'>Task 5</font> - Calculate and plot statistics

Until now we are working with hourly or daily data. These data with high temporal resolution are useful for studying short-term processes, such as changes between day and night or flux responses to individual weather events. However, when we are interested in seasonal patterns, long-term trends, or differences between months, the large amount of short-term variability can make the data difficult to interpret. Therefore, it is often practical to resample the data to a coarser time resolution, such as monthly values. We use ``resample`` function for pandas DataFrame. Check more details about this function on [pandas website](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.resample.html).

#### <font color='darkred'>**Example 1 (line plot)**</font>
Here's an example of calculating the monthly mean GPP for Hyltemossa station.

The code below will generate the following plot:
<br>
<img src="ancillarydata/images/exercise1/lineplot.png" width=40%>

````python
# Read data from Hyltemossa station:
htm_df = pd.read_csv("/data/project/climbeco/data/fluxnet/obs_dd_warm_winter/FLX_SE-Htm_FLUXNET2015_FULLSET_DD_2015_2020.csv",
                     header = 0,
                     sep = ",",
                     parse_dates = ["TIMESTAMP"])

# Set the "TIMESTAMP" column as index:
htm_df_ind = htm_df.set_index('TIMESTAMP')

# Calculate the mean GPP value for every month in the dataset:
monthly_mean_gpp = htm_df_ind.GPP_DT_VUT_MEAN.resample('ME').mean().dropna()

# Create line-plot:
monthly_mean_gpp.plot(kind='line', figsize=(10,6))
````

#### <font color='darkred'>**Example 2 (bar plot)**</font>
Here's an example of calculating the annual GPP and RECO for Hyltemossa station.

The code below will generate the following output:
<br>
<img src="ancillarydata/images/exercise1/plot2_task5.png" width=40%>

````python
# Read data from Hyltemossa station:
htm_df = pd.read_csv("/data/project/climbeco/data/fluxnet/obs_dd_warm_winter/FLX_SE-Htm_FLUXNET2015_FULLSET_DD_2015_2020.csv",
                     header = 0,
                     sep = ",",
                     parse_dates = ["TIMESTAMP"])

# Set the "TIMESTAMP" column as index:
htm_df_ind = htm_df.set_index('TIMESTAMP')

# Calculate annual means for GPP and RECO
# The variables named called '_MEAN' indicates that these columns contain the average 
# GPP and RECO values based on different methods.
annual_mean_GPP = htm_df_ind.GPP_DT_VUT_MEAN.resample('YE').mean().dropna()
annual_mean_RECO = htm_df_ind.RECO_DT_VUT_MEAN.resample('YE').mean().dropna()

# Define the width of the bars
bar_width = 0.4
# Create an array of indices for the x-axis based on the number of years
indices = np.arange(len(annual_mean_GPP))

# Plot the GPP bars in green
plt.bar(indices, annual_mean_GPP, width=bar_width, color='blue', label='GPP')

# Plot the RECO bars in red, shifted by the width of one bar to place them next to the GPP bars
plt.bar(indices + bar_width, annual_mean_RECO, width=bar_width, color='red', label='RECO')

# Customize the plot
plt.xlabel('Year')
plt.ylabel('CO$_2$ Flux (g C m$^{-2}$ d$^{-1}$)')
plt.title('Annual Mean of GPP and RECO')
plt.xticks(indices + bar_width / 2, annual_mean_GPP.index.year, rotation=45)  # Set x-axis labels to the years
plt.legend()
plt.grid(True)

# Display the plot
plt.tight_layout()  # Adjust layout to prevent clipping of tick-labels
plt.show()
````
